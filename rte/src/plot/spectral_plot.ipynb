```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import FuncFormatter
```

```python
def formatnum(x, pos):
    return f"2$^{int(np.log2(x))}$"


formatter = FuncFormatter(formatnum)
```

```python
Q = [8, 16, 32, 64, 128]
err = [
    0.0998722,
    0.00397611,
    0.00017662,
    1.04985497e-06,
    3.69059319e-10,
]  # rfm
# err = [
#     0.01795244,
#     1.11961393e-06,
#     2.35114997e-11,
#     2.70627193e-13,
#     4.22706134e-13,
# ]  # aprfm
```

```python
font_format = {"family": "Times New Roman", "size": 20}
plt.figure(figsize=(8, 6))
ax1 = plt.subplot(1, 1, 1)
plt.plot(
    Q,
    err,
    "-o",
    color="#94A3C0",
    markersize=8,
    linewidth=2,
)
plt.yscale("log")
# font size and type
plt.xticks(Q, fontproperties="Times New Roman", size=12)
plt.yticks(fontproperties="Times New Roman", size=12)
plt.xlabel("Degrees of freedom", font_format)
plt.ylabel("Error", font_format)
plt.grid()
# line width of four axises
ax = plt.gca()
ax.xaxis.set_major_formatter(formatter)
ax.spines["bottom"].set_linewidth(2)
ax.spines["left"].set_linewidth(2)
ax.spines["right"].set_linewidth(2)
ax.spines["top"].set_linewidth(2)
plt.savefig("./error_semilog_plot.png", dpi=300)
plt.show()
plt.close()
```
